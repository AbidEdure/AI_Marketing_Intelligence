```python
# Cell 1
import sys; sys.path.insert(0, "..")
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from src.preprocess import load_raw, engineer_features

df = engineer_features(load_raw())
print(df.shape)
print(df["campaign_accepted"].value_counts(normalize=True))

# Cell 2 — target balance per channel / tier / discount
for col in ["preferred_channel", "city_tier", "membership_tier", "campaign_type"]:
    display(df.groupby(col)["campaign_accepted"].agg(["mean", "count"]).sort_values("mean", ascending=False))

# Cell 3 — quick distributions
sns.histplot(df["total_spend_12m"], log_scale=True); plt.show()
sns.boxplot(data=df, x="campaign_accepted", y="campaign_discount_percent"); plt.show()

# Cell 4 — correlation heatmap on numerics
num = df.select_dtypes("number")
sns.heatmap(num.corr()[["campaign_accepted"]].sort_values("campaign_accepted"), annot=True); plt.show()
```
